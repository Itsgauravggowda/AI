```python
print("1BM24CS102 - Gaurav G Gowda")
room_a = input("Enter state of Room A (dirty/clean): ").lower()
room_b = input("Enter state of Room B (dirty/clean): ").lower()
vacuum = input("Enter vacuum cleaning position (a/b): ").lower()


print("\nInitial state:")
print("Room A :", room_a)
print("Room B :", room_b)
print("Vacuum position:", vacuum)

print("\n-- Agent Action --")


if vacuum == "a":
    if room_a == "dirty":
        print("Room A is dirty -> suck")
        room_a = "clean"
    else:
        print("Room A is clean -> Move right to Room B")
        vacuum = "b"

        if room_b == "dirty":
            print("Vacuum is in Room B -> suck")
            room_b = "clean"

elif vacuum == "b":
    if room_b == "dirty":
        print("Room B is dirty -> suck")
        room_b = "clean"
    else:
        print("Room B is clean -> Move left to Room A")
        vacuum = "a"

        if room_a == "dirty":
            print("Vacuum is in Room A -> suck")
            room_a = "clean"


print("\n-- Final State --")
print("Room A :", room_a)
print("Room B :", room_b)
print("Vacuum position:", vacuum)

```
